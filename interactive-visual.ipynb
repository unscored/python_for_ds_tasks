```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv('Titanic-Dataset.csv')

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

fig = px.histogram(df, x='Age', nbins=30, title='Distribution of Age', labels={'Age': 'Age'}, color_discrete_sequence=['#636EFA'])
fig.show()
```

```python
fig = px.scatter(df, x='Age', y='Fare', color='Survived', size='FamilySize', hover_data=['Name', 'Pclass'], labels={'Survived': 'Survived?', 'Age': 'Age', 'Fare': 'Fare'}, color_continuous_scale='RdYlGn')

fig.show()
```

```python
fig = px.box(df, x="Pclass", y="Age", color="Sex", title="Age Distribution by Class", labels={"Pclass": "Passenger Class", "Age": "Age"})

fig.show()
```

```python
df_hierarhy = df.groupby(['Pclass', 'Sex', 'Survived']).size().reset_index(name='count')

display(df_hierarhy)
fig = px.sunburst(df_hierarhy, path=['Pclass', 'Sex', 'Survived'], values='count', color='Survived', color_discrete_map={0: 'red', 1: 'green'}, title="Survival Breakdown by Class and Sex")
fig.show()
```

```python
numeric = df.select_dtypes(include=['number'])
correlation = numeric.corr()

fig = px.imshow(correlation, color_continuous_scale='RdBu_r', title='Correlation Matrix', aspect='auto', text_auto=True)
fig.show()
```
