```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('Titanic-Dataset.csv')

df['Age'].hist()

df
```

```python
df['Survived'].value_counts().plot(kind='bar')
```

```python
df['Age'].hist(bins=30, edgecolor='black')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
df.groupby('Pclass')['Survived'].mean().plot(kind='bar', color='skyblue', edgecolor='black')

plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.xticks(rotation=0)
plt.show()
```

```python
df.boxplot(column='Fare', by='Pclass', figsize=(14, 6))
plt.title('Fare Distribution by Passenger Class')
plt.suptitle('')
plt.ylabel('Fare (USD)')
plt.show()
```

```python
df.groupby(['Pclass', 'Sex'])['Age'].mean().unstack().plot(kind='line', marker='o')
plt.title('Average Age by Passenger Class and Sex')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.legend(loc='upper left', title='Sex')
plt.show()
```
